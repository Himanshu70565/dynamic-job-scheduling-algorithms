#### Creating Dataframe object for the jobs that need to be scheduled

```python
import pandas as pd

jobs_data = {
    "jobs": ["A", "B", "C", "D", "E"],
    "arrival_time": [0, 1, 2, 3, 4],
    "processing_time": [11, 29, 31, 1, 2],
    "due_date": [61, 45, 31, 33, 32]
}

df = pd.DataFrame(jobs_data)
```

#### First Come First Serve Dynamic Job Scheduling

```python
# Sorting jobs received as per there arrival time
df_sorted = df.sort_values(by='arrival_time')
df_sorted
```

```python
# Finish Time, Flow Time(Finish Time - Arrival Time), Tardiness(max(Finish Time - Due Date, 0))
df_sorted["finish_time"] = None
df_sorted["flow_time"] = None
df_sorted["tardiness"] = None

current_time = 0

# Iterating over the received jobs as per first come first serve jobs
for index, row in df_sorted.iterrows():
    finish_time = current_time + row.processing_time

    # Updating data in the dataframe for the calculated values
    df_sorted.loc[index, "finish_time"] = finish_time
    df_sorted.loc[index, "flow_time"] = finish_time - row.arrival_time
    df_sorted.loc[index, "tardiness"] = max(finish_time - row.due_date, 0)

    # Updating current time with the finish time of the current job
    current_time = finish_time

    # DEBUG LOG: Uncomment this line to see logs
    # print(f"Index: {index}, Job: {row.jobs}, Arrival Time: {row.arrival_time}, Finish time: {finish_time}, Flow time: {finish_time - row.arrival_time}")
```

```python
avg_tardiness = df_sorted['tardiness'].mean()
avg_flow_time = df_sorted['flow_time'].mean()
```

```python
print(f"Avg Tardiness: {avg_tardiness}, Avg Flow time: {avg_flow_time}")
```
